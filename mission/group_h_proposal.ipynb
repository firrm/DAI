## 2. Mission Title & Research Question

**Title:** *Predicting Short-Term Stock Return Direction via Technical Indicators, Market Regimes, and Alternative Signals*

**Research question:**  
Can next week stock and ETF return direction predictions be improved by combining traditional technical indicators with Hidden Markov Model based market regimes and less conventional features like volume behaviour or lunar cycles?


**Why it matters:**  
Short-term stock return prediction is difficult because financial markets are noisy and market behaviour can change quickly across different conditions. A model that performs well during stable periods may become much less reliable during unstable market conditions. This project looks at whether combining standard technical indicators with Hidden Markov Model based market regimes can improve next week return direction prediction. It also explores whether less conventional features, such as volume behavior and lunar cycle variables, contain any useful predictive information even if their effects are considered relatively weak on their own.

**Potential data quality issues:**
The dataset may contain missing values due to market holidays, asset-specific trading history, or rolling-window calculations. We also need to be aware of extreme outliers caused by stock splits or unusual market shocks.

The biggest risk in this setup is look-ahead bias. We have to be very strict about ensuring that feature calculations on day t do not accidentally include price information from day t+1. Furthermore, standard random cross-validation doesn't work well for time-series data due to autocorrelation. To prevent data leakage during training and testing, we are planing to use a purged and embargoed cross-validation approach.
