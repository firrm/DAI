## 2. Mission Title & Research Question

**Title:** *Predicting Loan Default Risk Using Machine Learning and Customer Segmentation*

**Research Question:**  
Can machine learning models accurately predict whether a borrower will default on a loan based on demographic, financial, and credit-related information, and can borrower segments be identified through clustering techniques to better understand default behavior?

**Why it matters:**  
Loan default prediction is a critical problem for financial institutions. Incorrect lending decisions can result in significant financial losses and increased credit risk. By combining predictive machine learning models with borrower segmentation techniques, financial institutions can make more informed lending decisions, improve risk management strategies, and reduce losses caused by high-risk borrowers.

## 3. Data

**Source(s):**

- Loan Default Dataset (provided for educational purposes)
- Format: CSV
- Access Method: Local dataset used within the project
- License: Academic/Educational Use

**Unit of Observation:**  
Each row represents a single loan application and its associated borrower information.

### Key Variables

| Variable | Type | Role | Description |
|-----------|------|------|-------------|
| Age | Numerical | Feature | Age of borrower |
| Income | Numerical | Feature | Annual income |
| Employment Length | Numerical | Feature | Years of employment |
| Loan Amount | Numerical | Feature | Amount borrowed |
| Credit Score | Numerical | Feature | Borrower's creditworthiness |
| Debt-to-Income Ratio | Numerical | Feature | Existing debt relative to income |
| Previous Defaults | Numerical | Feature | Number of historical defaults |
| Loan Purpose | Categorical | Feature | Reason for taking the loan |
| Default Status | Binary | Target | Indicates whether borrower defaulted |

### Potential Data Quality Issues

- Missing values in financial attributes
- Class imbalance between default and non-default cases
- Outliers in income and loan amount variables
- Potential correlation among financial indicators
- Possible sampling bias depending on data collection source

## 4. Planned Methods

Our project will apply techniques from all three required categories.

### 4a. Causal Inference

- [x] Causal graph / DAG (DoWhy)
- [x] Backdoor adjustment

**Justification:**

We will construct a causal graph to understand relationships between borrower characteristics, financial indicators, and default outcomes. This will help identify potential confounders and improve interpretation of predictive results beyond simple correlations.

---

### 4b. Supervised Learning

- [x] Logistic Regression
- [x] Decision Tree
- [x] Random Forest
- [x] Support Vector Machine

**Justification:**

Since loan default prediction is a binary classification problem, supervised learning techniques are suitable for predicting default outcomes. Multiple models will be compared to identify the most effective approach for classification performance and interpretability.

---

### 4c. Unsupervised Learning / Generative Models

- [x] K-Means Clustering

**Justification:**

K-Means clustering will be used to identify borrower groups with similar financial and behavioral characteristics. These segments may reveal hidden patterns associated with loan default risk and provide additional business insights.

## 5. Evaluation Strategy

### Classification Models

The following metrics will be used:

- Accuracy
- Precision
- Recall
- F1 Score
- ROC-AUC Score
- Confusion Matrix

Special emphasis will be placed on Recall because failing to identify a risky borrower may have significant financial consequences.

### Clustering Evaluation

For K-Means clustering:

- Silhouette Score
- Cluster Visualization
- Cluster Interpretation

### Causal Evaluation

For causal analysis:

- DAG validation
- Refutation tests
- Sensitivity analysis
- Comparison with domain knowledge

### Baselines

The following baseline models will be used:

- Majority Class Prediction
- Logistic Regression

More advanced models will be compared against these baselines.
