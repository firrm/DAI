## 2. Mission Title & Research Question

**Title:** Forecasting Financial Market Volatility: Integrating Machine Learning, Causal Inference, and Market Regimes

**Research question:** To what extent does a supervised machine learning model (XGBoost) outperform traditional econometric baselines (GARCH) in forecasting the realized volatility of the DAX index, and how do latent, unsupervised market regimes and causal structural shocks affect this predictive performance?

**Why it matters:** Accurate volatility forecasting is the cornerstone of modern risk management, portfolio optimization, and derivative pricing. By bridging traditional econometrics with modern AI tools—and structurally understanding the causal drivers of market shocks—policymakers and financial institutions can better anticipate and mitigate the impacts of severe market downturns.

## 3. Data

**Source(s):** Historical market data for the DAX Index (^GDAXI) spanning January 1, 2015, to January 1, 2026. The primary data source will be WRDS (Wharton Research Data Services) or alternatively the Yahoo Finance API (yfinance). Licence: Open access for academic/research purposes.

**Unit of observation:** One trading day.

**Key variables:**

## 5. Evaluation Strategy

*How will you know if your mission succeeded? Describe:*

- **Supervised Metrics:** We will use Mean Absolute Error (MAE) and Root Mean Squared Error (RMSE) on a strict sequential 20% out-of-sample test set (to prevent temporal data leakage). 
- **Causal Validation:** We will validate our causal claims using DoWhy's refutation tests, specifically adding a random common cause and using placebo treatments to ensure our estimated causal effect is robust against unobserved confounding.
- **Baselines/Benchmarks:** The machine learning performance will be rigorously benchmarked against the classical **GARCH(1,1)** econometric model. Success is defined by XGBoost achieving a statistically significant reduction in out-of-sample RMSE compared to the GARCH baseline, while the clustering successfully stratifies high vs. low error distributions.
