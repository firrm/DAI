## 2. Mission Title & Research Question

**Title:** *Customer Churn in Energy Markets: Predicting Customer Churn and Estimating the Effect of Off-Peak Electricity Prices*

**Research question:**  
*To what extent does the average variable off-peak electricity price causally influence customer churn probability in the PowerCo customer dataset, after controlling for customer tenure, energy consumption, profitability, and contract characteristics, and how accurately can machine learning models identify customers at risk of churning?*

**Why it matters:**  
*Customer retention is a major challenge in competitive energy markets. Losing customers can reduce revenue, increase acquisition costs, and weaken a company's market position. One important economic question is whether energy pricing influences customers' decisions to remain with or leave a provider. Understanding the causal effect of electricity prices on customer churn can help energy companies design pricing strategies that balance profitability and customer retention. In addition, accurately identifying customers at risk of churning enables firms to target retention efforts more effectively. This project combines machine learning and causal inference to study both the prediction and the economic drivers of customer churn.*

### 4a. Causal Inference
- [+] Causal graph / DAG (DoWhy)
- [+] Backdoor adjustment
- [ ] Instrumental variable
- [ ] Propensity score stratification
- [ ] Other: ___

Treatment variable:

The treatment variable will be avg_price_off_peak_var, defined as the customer-level average variable off-peak electricity price aggregated from the monthly pricing dataset.

Outcome variable:

The outcome variable will be churn, indicating whether a customer left the company (1) or stayed (0).

Justification:

The causal inference component focuses on a specific economic mechanism: whether higher average variable off-peak electricity prices increase the probability of customer churn.

A Directed Acyclic Graph (DAG) will be constructed to represent the assumed relationships between off-peak electricity prices, customer characteristics, consumption behavior, profitability, contract characteristics, and churn.

Based on the DAG, backdoor adjustment will be used to control for observed confounding variables such as customer tenure (num_years_antig), electricity and gas consumption (cons_12m, cons_gas_12m), forecasted consumption, customer profitability (net_margin), gas contract status (has_gas), and sales channel (channel_sales).

This approach allows us to estimate the causal effect of the treatment variable avg_price_off_peak_var on customer churn under explicit causal assumptions.

We do not plan to use an instrumental variable approach because the dataset does not provide a clearly valid instrument. Instead, the causal analysis will rely on DAG-based backdoor adjustment and will carefully discuss limitations such as unobserved competitor prices, customer satisfaction, and regional market conditions.

The robustness of the estimated causal effect will be assessed using DoWhy refutation tests, including placebo treatment tests and random common cause tests.
