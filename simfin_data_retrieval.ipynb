## 9. Compute 1-Year Forward Returns

```python
price_col = 'Adj. Close' if 'Adj. Close' in df_prices.columns else 'Close'
prices_clean = df_prices[[price_col]].rename(columns={price_col: 'Price'})
prices_clean = prices_clean[prices_clean['Price'] > 0]

# Pre-build dict: ticker → Date-only index (Ticker level dropped)
prices_grouped = {
    ticker: grp.droplevel('Ticker').sort_index()
    for ticker, grp in prices_clean.groupby(level='Ticker')
}

def lookup_price(tp, target_date, max_lag=5):
    window = tp.loc[target_date : target_date + pd.Timedelta(days=max_lag)]
    return float(window['Price'].iloc[0]) if len(window) else np.nan

print("Computing 1-year forward returns...")
results = {}
for ticker, report_date in df_fund.index:
    tp = prices_grouped.get(str(ticker))
    if tp is None:
        results[(ticker, report_date)] = np.nan
        continue
    p0 = lookup_price(tp, report_date)
    p1 = lookup_price(tp, report_date + pd.Timedelta(days=252))
    results[(ticker, report_date)] = (p1 - p0) / p0 if (p0 and p1 and p0 != 0) else np.nan

df_fund['Perform'] = pd.Series(results)
print(f"Forward returns: {df_fund['Perform'].notna().sum():,} non-null / {len(df_fund):,} total")
print(df_fund['Perform'].describe().round(3))
```

## 10. Construct Class Labels

```python
import pandas_datareader.data as web

# ── Reset index for column-based operations ───────────────────────────────────
df_fund = df_fund.reset_index()
df_fund['Report Date'] = pd.to_datetime(df_fund['Report Date'])
df_fund['FiscalYear']  = df_fund['Report Date'].dt.year

# ────────────────────────────────────────────────────────────────────────────
# A. NBER Recession Indicator (Revision 1 — Macro Controls)
# ────────────────────────────────────────────────────────────────────────────
# Binary flag: 1 if the report date falls within an NBER-designated recession.
# Recession periods sourced from https://www.nber.org/research/data/us-business-cycle-expansions-and-contractions
# Current dataset covers FY2020-2024; only the COVID-19 recession is relevant.
NBER_RECESSIONS = [
    ('2020-02-01', '2020-04-30'),  # COVID-19 recession (NBER: Feb–Apr 2020)
]

def is_recession(date):
    for s, e in NBER_RECESSIONS:
        if pd.Timestamp(s) <= date <= pd.Timestamp(e):
            return 1
    return 0

df_fund['NBER_Recession'] = df_fund['Report Date'].apply(is_recession)
print(f"NBER_Recession — flagged rows: {df_fund['NBER_Recession'].sum():,}")

# ────────────────────────────────────────────────────────────────────────────
# B. InvestmentGrowth (Revision 4 — CMA Factor Proxy)
# ────────────────────────────────────────────────────────────────────────────
# Year-over-year percentage change in Total Assets.
# Proxy for the Fama-French CMA (Conservative Minus Aggressive) factor:
# firms with lower asset growth (conservative) earn higher returns.
df_fund = df_fund.sort_values(['Ticker', 'Report Date'])
df_fund['InvestmentGrowth'] = (
    df_fund.groupby('Ticker')['Total Assets']
    .pct_change()
    .clip(-2, 5)   # winsorise: -200% to +500% (avoids division near-zero edge cases)
)
print(f"InvestmentGrowth — non-null: {df_fund['InvestmentGrowth'].notna().sum():,}")

# ────────────────────────────────────────────────────────────────────────────
# C. MarketCap, Size (SMB Proxy), BookToMarket (HML Proxy) (Revision 4)
# ────────────────────────────────────────────────────────────────────────────
# MarketCap: Shares Outstanding × Close Price at Report Date.
# Size      = log(MarketCap) — SMB proxy (larger firms earn lower returns).
# BookToMarket = Total Equity / MarketCap — HML proxy (high BtM = value firms).
price_col = 'Adj. Close' if 'Adj. Close' in df_prices.columns else 'Close'
prices_clean = df_prices[[price_col, 'Shares Outstanding']].copy()
prices_clean.columns = ['Price', 'Shares']
prices_clean = prices_clean[(prices_clean['Price'] > 0) & prices_clean['Price'].notna()]

prices_by_ticker = {
    ticker: grp.droplevel('Ticker').sort_index()
    for ticker, grp in prices_clean.groupby(level='Ticker')
}

def lookup_mktcap(ticker, report_date, max_lag=5):
    tp = prices_by_ticker.get(str(ticker))
    if tp is None:
        return np.nan
    window = tp.loc[report_date : report_date + pd.Timedelta(days=max_lag)]
    if len(window) == 0:
        return np.nan
    row = window.iloc[0]
    price  = row['Price']
    shares = row['Shares']
    if pd.isna(price) or pd.isna(shares) or shares <= 0:
        return np.nan
    return price * shares

print("Computing MarketCap (Size + BookToMarket proxies)...")
mktcap_values = [
    lookup_mktcap(row['Ticker'], row['Report Date'])
    for _, row in df_fund.iterrows()
]
df_fund['MarketCap']    = mktcap_values
df_fund['Size']         = np.log(df_fund['MarketCap'].clip(lower=1))
df_fund['BookToMarket'] = df_fund['Total Equity'] / df_fund['MarketCap'].replace(0, np.nan)
print(f"MarketCap — non-null: {df_fund['MarketCap'].notna().sum():,}")
print(f"Size      — non-null: {df_fund['Size'].notna().sum():,}")
print(f"BookToMarket — non-null: {df_fund['BookToMarket'].notna().sum():,}")

# ────────────────────────────────────────────────────────────────────────────
# D. Survivorship Bias Statistics (Revision 6)
# ────────────────────────────────────────────────────────────────────────────
# Report unique-ticker counts per fiscal year.
# Compare against Compustat/CRSP benchmark (~8,000-10,000 US firms per year).
# SimFin free tier covers only currently-listed firms — delisted/bankrupt absent.
print("\n=== Survivorship Bias Check — Unique Tickers per Fiscal Year ===")
print("(SimFin free tier; compare against Compustat ~8,000-10,000 for magnitude of bias)")
print(df_fund.groupby('FiscalYear')['Ticker'].nunique().to_string())
print("\nNote: Missing firms are concentrated in the left tail of returns (delisted = poor performers).")
print("This inflates apparent Buy-signal performance. All results qualified accordingly.")

# ────────────────────────────────────────────────────────────────────────────
# E. Calendar-Quarter Cohort Labeling (Revision 2 — Target Variable Alignment)
# ────────────────────────────────────────────────────────────────────────────
# Original: ranked within fiscal year → mixes firms with staggered fiscal year-ends
#           (e.g., FY ending Jan vs. Dec), creating non-synchronised return windows.
# Revised:  assign each firm to the calendar quarter when its report becomes public
#           (Report Date + 90 days disclosure lag), then rank within that cohort.
#           All firms in a cohort share synchronised, non-overlapping 252-day windows.

def _tercile(group):
    r = group['Perform'].rank(method='first', na_option='keep', pct=True)
    return pd.cut(r, bins=[0, 1/3, 2/3, 1.0],
                  labels=[-1, 0, 1], include_lowest=True).astype(float)

df_fund['PublicDate']      = df_fund['Report Date'] + pd.Timedelta(days=90)
df_fund['CalendarQuarter'] = (
    df_fund['PublicDate'].dt.year.astype(str) + '-Q' +
    df_fund['PublicDate'].dt.quarter.astype(str)
)

# Only assign class where forward return is available
df_with_return = df_fund[df_fund['Perform'].notna()].copy()
df_with_return['Class'] = (
    df_with_return
    .groupby('CalendarQuarter', group_keys=False)
    .apply(_tercile, include_groups=False)
)

# Merge class labels back (left join — rows without return get NaN class)
df_fund = df_fund.merge(
    df_with_return[['Ticker', 'Report Date', 'Class']],
    on=['Ticker', 'Report Date'],
    how='left'
)

print("\nClass distribution (calendar-quarter cohort labeling):")
print(df_fund['Class'].value_counts().sort_index())
print(f"\nCalendar quarters with labelled data: {df_fund.loc[df_fund['Class'].notna(), 'CalendarQuarter'].nunique()}")
print(df_fund.groupby('CalendarQuarter')['Class'].count().head(8).to_string())

# ────────────────────────────────────────────────────────────────────────────
# F. Fama-French 5-Factor Download (Revision 5 — FF3 → FF5)
# ────────────────────────────────────────────────────────────────────────────
# Download monthly FF5 factors (MKT-RF, SMB, HML, RMW, CMA) from the
# Kenneth French Data Library.  Saved to ff5_factors.csv for use in:
#   (1) economic FF5 baseline classifier, and
#   (2) Long-Short portfolio alpha regression.
try:
    ff5 = web.DataReader(
        'F-F_Research_Data_5_Factors_2x3', 'famafrench',
        start='2019-01-01'
    )[0]
    ff5 = ff5 / 100   # convert percentage to decimal
    ff5.index = pd.to_datetime(ff5.index.to_timestamp())  # PeriodIndex → DatetimeIndex
    ff5.to_csv('ff5_factors.csv')
    print(f"\nFF5 factors downloaded: {ff5.shape[0]} monthly observations")
    print(ff5.tail(3))
except Exception as e:
    print(f"\nFF5 download failed: {e}")
    print("Install pandas-datareader: pip install pandas-datareader")
    print("If offline, place ff5_factors.csv manually from https://mba.tuck.dartmouth.edu/pages/faculty/ken.french/data_library.html")
```
